## About outliers in data analysis:

**Common methods for detecting outliers**

Boxplots:

What it does: Visualizes the distribution of data and identifies outliers as points outside the “whiskers” (typically 1.5 times the interquartile range).

When to use: For a quick, visual inspection of outliers in a single variable.

Z-Scores:

What it does: Measures how many standard deviations a data point is from the mean. Data points with Z-scores greater than 3 or less than -3 are often considered outliers.

When to use: For numerical data when you want a statistical measure of outliers.

IQR (Interquartile Range) Method:

What it does: The Interquartile Range (IQR) method detects outliers by finding data points that fall below Q₁ - 1.5 x IQR or above Q₃ + 1.5 x IQR.

When to use: For robust outlier detection, especially when the data is not normally distributed.

**Handling Outliers**

Winsorization: Replaced with the nearest non-outlier value.

Removing: Excluded from the dataset if they are errors or irrelevant.

Keeping: Retained if they represent valid, meaningful observations.

Proper outlier detection ensures your model is not unduly influenced by extreme values, leading to more reliable and accurate results.

Reference: Data Science & Analytics in Finance, Economics & Business course book by Mikko Ranta [Link]

## Dealing with outliers

Extreme values may bias the analysis, so this time we will winsorize the data by setting cutoffs at 2.5% and 97.5%. Make sure to not winsorize for example industry codes.

```python
#Winsorize
df[num_cols] = df[num_cols].clip(lower=df[num_cols].quantile(0.025), upper=df[num_cols].quantile(0.975), axis=1)
```

## Industry comparison

Let's divide the dataset into two groups: restaurants and hotels.
Then we will take another look at the descriptive statistics and histograms to compare the industries.

```python
#55=hotel 56=restaurant

df['short_industry'] = df["NACE Rev. 2, core code (4 digits)"].astype(str).str[:2].astype(int)
df["Industry"] = df["short_industry"].map({55: "Hotel", 56: "Restaurant"})
df[["NACE Rev. 2, core code (4 digits)", 'short_industry', "Industry"]]
```

```python
df[['Profit margin Last avail. yr','ROE using P/L before tax Last avail. yr',
    'ROA using Net income Last avail. yr', 'Total assets EUR Last avail. yr', "Industry"]].groupby("Industry").describe().T
#Max values are equal due to winsorization
```
